```python

# ==========================================
# EMPLOYEE RETENTION INTELLIGENCE SYSTEM
# 01 - DATA GENERATION
# ==========================================

# Import libraries
import pandas as pd
import numpy as np
import random
import os

from datetime import datetime, timedelta

# Set random seeds for reproducibility
np.random.seed(42)
random.seed(42)

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 20)

print("Libraries imported successfully!")
print("Employee Retention Intelligence Project")
```

```python

# ==========================================
# CELL 2: ORGANISATIONAL CONFIGURATION
# ==========================================

# Fictional organisation details
COMPANY_NAME = "Nexus Workforce Solutions"
TOTAL_EMPLOYEES = 2000

# Organisational departments
departments = [
    "Human Resources",
    "Finance",
    "Information Technology",
    "Sales",
    "Marketing",
    "Operations",
    "Customer Support",
    "Business Development",
    "Legal",
    "Administration"
]

# Job roles within the organisation
job_roles = {
    "Human Resources": [
        "HR Analyst", "Recruitment Officer",
        "HR Manager", "Payroll Specialist"
    ],
    "Finance": [
        "Accountant", "Financial Analyst",
        "Finance Manager", "Auditor"
    ],
    "Information Technology": [
        "Data Analyst", "Software Developer",
        "IT Support Officer", "IT Manager"
    ],
    "Sales": [
        "Sales Executive", "Sales Manager",
        "Account Manager", "Sales Analyst"
    ],
    "Marketing": [
        "Marketing Executive", "Brand Manager",
        "Content Strategist", "Marketing Analyst"
    ],
    "Operations": [
        "Operations Analyst", "Operations Manager",
        "Process Coordinator", "Logistics Officer"
    ],
    "Customer Support": [
        "Customer Service Officer", "Support Analyst",
        "Team Lead", "Customer Experience Manager"
    ],
    "Business Development": [
        "Business Development Executive",
        "Partnership Manager",
        "Business Analyst",
        "Growth Manager"
    ],
    "Legal": [
        "Legal Officer", "Compliance Analyst",
        "Legal Manager", "Contract Specialist"
    ],
    "Administration": [
        "Administrative Officer", "Office Manager",
        "Executive Assistant", "Facilities Officer"
    ]
}

# Office locations
locations = [
    "Lagos",
    "Abuja",
    "Port Harcourt",
    "Ibadan",
    "Kano",
    "Enugu"
]

# Employment categories
employment_types = [
    "Full-time",
    "Contract",
    "Part-time"
]

# Gender categories for synthetic data
genders = [
    "Male",
    "Female"
]

# Organisational employment grades
job_grades = [
    "Entry Level",
    "Junior",
    "Mid-Level",
    "Senior",
    "Management"
]

# Reporting period
START_DATE = pd.Timestamp("2024-01-01")
END_DATE = pd.Timestamp("2025-12-31")

# Display organisation configuration
print("Organisation:", COMPANY_NAME)
print("Total Employees:", TOTAL_EMPLOYEES)
print("Number of Departments:", len(departments))
print("Number of Locations:", len(locations))
print("Reporting Period:", START_DATE.date(), "to", END_DATE.date())
```

```python

# ==========================================
# CELL 3: GENERATE EMPLOYEE MASTER DATA
# ==========================================

# Generate employee IDs
employee_ids = [
    f"NWS{str(i).zfill(4)}"
    for i in range(1, TOTAL_EMPLOYEES + 1)
]

# Define salary ranges by job grade (monthly NGN)
salary_ranges = {
    "Entry Level": (120000, 220000),
    "Junior": (180000, 350000),
    "Mid-Level": (300000, 600000),
    "Senior": (500000, 900000),
    "Management": (800000, 1800000)
}

# Probability distribution for job grades
grade_weights = [0.15, 0.25, 0.30, 0.20, 0.10]

# Department distribution
department_weights = [
    0.10, 0.08, 0.12, 0.13, 0.08,
    0.16, 0.12, 0.09, 0.05, 0.07
]

# Generate employee records
employees = []

for emp_id in employee_ids:

    department = random.choices(
        departments,
        weights=department_weights,
        k=1
    )[0]

    job_role = random.choice(job_roles[department])

    grade = random.choices(
        job_grades,
        weights=grade_weights,
        k=1
    )[0]

    age = int(np.random.normal(34, 7))
    age = max(21, min(age, 60))

    gender = random.choices(
        genders,
        weights=[0.52, 0.48],
        k=1
    )[0]

    location = random.choice(locations)

    employment_type = random.choices(
        employment_types,
        weights=[0.80, 0.15, 0.05],
        k=1
    )[0]

    # Generate salary based on grade
    min_salary, max_salary = salary_ranges[grade]

    salary = random.randint(min_salary, max_salary)

    # Generate employment date
    hire_date = START_DATE + pd.to_timedelta(
        random.randint(0, (END_DATE - START_DATE).days),
        unit="D"
    )

    # Previous work experience
    previous_experience = max(
        0,
        round(np.random.normal(4, 2), 1)
    )

    # Educational qualification
    education = random.choices(
        ["OND/NCE", "BSc/HND", "MSc", "PhD"],
        weights=[0.10, 0.55, 0.30, 0.05],
        k=1
    )[0]

    employees.append({
        "Employee_ID": emp_id,
        "Age": age,
        "Gender": gender,
        "Department": department,
        "Job_Role": job_role,
        "Job_Grade": grade,
        "Location": location,
        "Employment_Type": employment_type,
        "Monthly_Salary": salary,
        "Hire_Date": hire_date,
        "Previous_Experience_Years": previous_experience,
        "Education": education
    })

# Convert records into DataFrame
employee_df = pd.DataFrame(employees)

# Display first five records
display(employee_df.head())

# Check dataset dimensions
print("Dataset shape:", employee_df.shape)

# Check for duplicate employee IDs
print(
    "Duplicate Employee IDs:",
    employee_df["Employee_ID"].duplicated().sum()
)
```

```python

# ==========================================
# CELL 4: PERFORMANCE & WORKPLACE EXPERIENCE
# ==========================================

experience_records = []

for _, employee in employee_df.iterrows():

    # Performance rating
    performance = np.random.choice(
        [1, 2, 3, 4, 5],
        p=[0.05, 0.15, 0.35, 0.30, 0.15]
    )

    # KPI achievement
    kpi = np.random.normal(75 + (performance - 3) * 6, 10)
    kpi = round(np.clip(kpi, 20, 120), 1)

    # Job satisfaction
    satisfaction = np.random.choice(
        [1, 2, 3, 4, 5],
        p=[0.08, 0.15, 0.32, 0.30, 0.15]
    )

    # Work-life balance
    work_life = np.random.choice(
        [1, 2, 3, 4, 5],
        p=[0.08, 0.15, 0.30, 0.32, 0.15]
    )

    # Engagement score
    engagement = np.random.normal(
        65 + (satisfaction - 3) * 8,
        12
    )
    engagement = round(np.clip(engagement, 0, 100), 1)

    # Manager relationship
    manager_relationship = np.random.choice(
        [1, 2, 3, 4, 5],
        p=[0.08, 0.12, 0.30, 0.32, 0.18]
    )

    # Workload score
    workload = np.random.normal(
        60 + (3 - work_life) * 7,
        12
    )
    workload = round(np.clip(workload, 0, 100), 1)

    # Overtime hours per month
    overtime = np.random.gamma(
        shape=2,
        scale=4
    )
    overtime = round(np.clip(overtime, 0, 40), 1)

    # Training hours per year
    training = int(np.random.gamma(3, 8))
    training = min(training, 120)

    # Employee tenure
    tenure_years = (
        END_DATE - employee["Hire_Date"]
    ).days / 365.25

    # Time in current role
    years_in_role = round(
        np.random.uniform(0, max(0.1, tenure_years)),
        1
    )

    # Time since last promotion
    years_since_promotion = round(
        np.random.uniform(0, max(0.1, tenure_years)),
        1
    )

    experience_records.append({
        "Employee_ID": employee["Employee_ID"],
        "Performance_Rating": performance,
        "KPI_Achievement": kpi,
        "Job_Satisfaction": satisfaction,
        "Work_Life_Balance": work_life,
        "Engagement_Score": engagement,
        "Manager_Relationship": manager_relationship,
        "Workload_Score": workload,
        "Overtime_Hours": overtime,
        "Training_Hours": training,
        "Years_In_Current_Role": years_in_role,
        "Years_Since_Promotion": years_since_promotion
    })

# Convert to DataFrame
experience_df = pd.DataFrame(experience_records)

# Merge with employee master dataset
employee_df = employee_df.merge(
    experience_df,
    on="Employee_ID",
    how="left"
)

# Display results
display(employee_df.head())

print("Updated dataset shape:", employee_df.shape)

# Check missing values
print("\nMissing values:")
print(employee_df.isnull().sum())
```

```python

# ==========================================
# CELL 5: CAREER PROGRESSION & COMPENSATION
# ==========================================

career_records = []

for _, employee in employee_df.iterrows():

    tenure = (
        END_DATE - employee["Hire_Date"]
    ).days / 365.25

    # Number of promotions based on tenure
    promotion_count = np.random.poisson(
        max(0.1, tenure / 3)
    )

    promotion_count = min(promotion_count, 5)

    # Internal role changes
    mobility_count = np.random.poisson(
        max(0.1, tenure / 4)
    )

    mobility_count = min(mobility_count, 5)

    # Salary increment percentage
    increment = round(
        np.random.normal(8, 5),
        2
    )

    increment = max(-5, min(increment, 30))

    # Previous salary
    previous_salary = round(
        employee["Monthly_Salary"] / (1 + increment / 100)
    )

    # Last promotion date
    if promotion_count > 0:
        days_since_hire = (
            END_DATE - employee["Hire_Date"]
        ).days

        promotion_date = (
            employee["Hire_Date"] +
            pd.to_timedelta(
                random.randint(0, days_since_hire),
                unit="D"
            )
        )
    else:
        promotion_date = pd.NaT

    # Career development score
    career_score = np.random.normal(
        55 + promotion_count * 7 +
        mobility_count * 5,
        12
    )

    career_score = round(
        np.clip(career_score, 0, 100),
        1
    )

    # Benefits satisfaction
    benefits = np.random.choice(
        [1, 2, 3, 4, 5],
        p=[0.08, 0.15, 0.30, 0.32, 0.15]
    )

    # Salary competitiveness
    salary_competitiveness = np.random.choice(
        [1, 2, 3, 4, 5],
        p=[0.12, 0.18, 0.30, 0.25, 0.15]
    )

    career_records.append({
        "Employee_ID": employee["Employee_ID"],
        "Previous_Salary": previous_salary,
        "Salary_Increment_Percent": increment,
        "Last_Promotion_Date": promotion_date,
        "Promotion_Count": promotion_count,
        "Internal_Mobility_Count": mobility_count,
        "Career_Development_Score": career_score,
        "Benefits_Satisfaction": benefits,
        "Salary_Competitiveness": salary_competitiveness
    })

# Create career progression DataFrame
career_df = pd.DataFrame(career_records)

# Merge with employee master dataset
employee_df = employee_df.merge(
    career_df,
    on="Employee_ID",
    how="left"
)

# Display first five records
display(employee_df.head())

# Dataset dimensions
print("Updated dataset shape:", employee_df.shape)

# Summary statistics
display(
    employee_df[
        [
            "Monthly_Salary",
            "Previous_Salary",
            "Salary_Increment_Percent",
            "Promotion_Count",
            "Internal_Mobility_Count",
            "Career_Development_Score"
        ]
    ].describe().round(2)
)
```

```python

# ==========================================
# CELL 6: GENERATE EMPLOYEE ATTRITION RECORDS
# ==========================================

exit_records = []

# Create employment status column
employee_df["Employment_Status"] = "Active"

for index, employee in employee_df.iterrows():

    # Calculate length of employment exposure
    tenure_days = (
        END_DATE - employee["Hire_Date"]
    ).days

    exposure_years = tenure_days / 365.25

    # Calculate employee retention risk factors
    satisfaction_risk = (5 - employee["Job_Satisfaction"]) / 4

    engagement_risk = (
        100 - employee["Engagement_Score"]
    ) / 100

    work_life_risk = (
        5 - employee["Work_Life_Balance"]
    ) / 4

    compensation_risk = (
        5 - employee["Salary_Competitiveness"]
    ) / 4

    career_risk = (
        5 - employee["Benefits_Satisfaction"]
    ) / 4

    promotion_risk = min(
        employee["Years_Since_Promotion"] / 5,
        1
    )

    workload_risk = employee["Workload_Score"] / 100

    # Combine risk factors
    risk_score = (
        0.20 * satisfaction_risk +
        0.20 * engagement_risk +
        0.15 * work_life_risk +
        0.15 * compensation_risk +
        0.10 * career_risk +
        0.10 * promotion_risk +
        0.10 * workload_risk
    )

    # Convert risk score into an annual exit hazard
    annual_hazard = 0.10 * np.exp(
        2 * (risk_score - 0.5)
    )

    # Probability of leaving during observed tenure
    exit_probability = 1 - np.exp(
        -annual_hazard * exposure_years
    )

    # Generate resignation outcome
    resigned = np.random.random() < exit_probability

    if resigned and tenure_days >= 30:

        # Generate exit date
        days_to_exit = random.randint(
            30,
            tenure_days
        )

        exit_date = (
            employee["Hire_Date"] +
            pd.to_timedelta(days_to_exit, unit="D")
        )

        # Possible resignation reasons
        reasons = [
            "Better Career Opportunity",
            "Inadequate Compensation",
            "Limited Career Progression",
            "Work-Life Balance",
            "Management Issues",
            "Personal Reasons",
            "Further Education"
        ]

        # Resignation reasons influenced by employee experience
        reason_weights = [
            0.15,
            0.10 + compensation_risk * 0.15,
            0.10 + promotion_risk * 0.20,
            0.10 + work_life_risk * 0.15,
            0.10 + satisfaction_risk * 0.10,
            0.15,
            0.10
        ]

        reason_weights = np.array(reason_weights)
        reason_weights = reason_weights / reason_weights.sum()

        exit_reason = random.choices(
            reasons,
            weights=reason_weights,
            k=1
        )[0]

        employee_df.loc[
            index, "Employment_Status"
        ] = "Resigned"

        exit_records.append({
            "Employee_ID": employee["Employee_ID"],
            "Exit_Date": exit_date,
            "Exit_Type": "Voluntary",
            "Exit_Reason": exit_reason
        })

# Create exit records DataFrame
exit_df = pd.DataFrame(exit_records)

# Display exit records
display(exit_df.head())

# Display attrition summary
attrition_count = len(exit_df)

active_count = (
    employee_df["Employment_Status"] == "Active"
).sum()

attrition_rate = (
    attrition_count / TOTAL_EMPLOYEES
) * 100

print("Total Employees:", TOTAL_EMPLOYEES)
print("Employees Resigned:", attrition_count)
print("Employees Active:", active_count)
print("Historical Attrition Rate:", round(attrition_rate, 2), "%")

# Exit reason distribution
if not exit_df.empty:
    print("\nExit Reasons:")
    display(exit_df["Exit_Reason"].value_counts())
```

```python

# ==========================================
# CELL 7: GENERATE MONTHLY WORKFORCE HISTORY
# ==========================================

monthly_records = []

# Generate 24 month-end observation dates
observation_dates = pd.date_range(
    start=START_DATE,
    end=END_DATE,
    freq="ME"
)

# Create employee exit date lookup
exit_lookup = {}

if not exit_df.empty:
    exit_lookup = dict(
        zip(exit_df["Employee_ID"], exit_df["Exit_Date"])
    )

# Last date with a fully observable six-month outcome
last_valid_observation = (
    END_DATE - pd.DateOffset(months=6)
)

for _, employee in employee_df.iterrows():

    employee_id = employee["Employee_ID"]
    hire_date = employee["Hire_Date"]

    exit_date = exit_lookup.get(
        employee_id,
        pd.NaT
    )

    for observation_date in observation_dates:

        # Employee must have joined by observation date
        if hire_date > observation_date:
            continue

        # Exclude employees who have already exited
        if pd.notna(exit_date) and exit_date <= observation_date:
            continue

        month_number = (
            (observation_date.year - hire_date.year) * 12
            + observation_date.month - hire_date.month
        )

        # Simulate gradual changes in employee experience
        engagement = np.clip(
            employee["Engagement_Score"]
            + np.random.normal(0, 5),
            0, 100
        )

        satisfaction = int(np.clip(
            round(
                employee["Job_Satisfaction"]
                + np.random.normal(0, 0.5)
            ),
            1, 5
        ))

        work_life = int(np.clip(
            round(
                employee["Work_Life_Balance"]
                + np.random.normal(0, 0.5)
            ),
            1, 5
        ))

        workload = np.clip(
            employee["Workload_Score"]
            + np.random.normal(0, 6),
            0, 100
        )

        overtime = np.clip(
            employee["Overtime_Hours"]
            + np.random.normal(0, 3),
            0, 40
        )

        manager_relationship = int(np.clip(
            round(
                employee["Manager_Relationship"]
                + np.random.normal(0, 0.4)
            ),
            1, 5
        ))

        # Calculate employee tenure at observation date
        tenure = round(
            (observation_date - hire_date).days / 365.25,
            2
        )

        # Calculate six-month attrition target
        if observation_date <= last_valid_observation:

            target = int(
                pd.notna(exit_date)
                and observation_date < exit_date
                <= observation_date + pd.DateOffset(months=6)
            )

        else:
            # Outcome is not fully observable
            target = np.nan

        monthly_records.append({

            "Employee_ID": employee_id,
            "Observation_Date": observation_date,
            "Months_At_Company": max(0, month_number),
            "Tenure_Years": tenure,

            "Monthly_Salary": employee["Monthly_Salary"],
            "Performance_Rating": employee["Performance_Rating"],
            "KPI_Achievement": employee["KPI_Achievement"],

            "Job_Satisfaction": satisfaction,
            "Work_Life_Balance": work_life,
            "Engagement_Score": round(engagement, 1),
            "Manager_Relationship": manager_relationship,

            "Workload_Score": round(workload, 1),
            "Overtime_Hours": round(overtime, 1),
            "Training_Hours": employee["Training_Hours"],

            "Years_In_Current_Role": employee["Years_In_Current_Role"],
            "Years_Since_Promotion": employee["Years_Since_Promotion"],
            "Career_Development_Score": employee["Career_Development_Score"],

            "Salary_Competitiveness": employee["Salary_Competitiveness"],

            "Attrition_Within_6M": target

        })

# Convert into DataFrame
monthly_df = pd.DataFrame(monthly_records)

# Display sample records
display(monthly_df.head())

# Dataset overview
print("Monthly Workforce Dataset")
print("-------------------------")
print("Total Records:", len(monthly_df))
print("Unique Employees:", monthly_df["Employee_ID"].nunique())
print("Observation Period:",
      monthly_df["Observation_Date"].min(),
      "to",
      monthly_df["Observation_Date"].max())

# Check prediction target distribution
print("\nSix-Month Attrition Target:")
display(
    monthly_df["Attrition_Within_6M"]
    .value_counts(dropna=False)
    .rename("Record_Count")
)

# Check missing values
print("\nMissing Values:")
display(monthly_df.isnull().sum())
```

```python

# ==========================================
# CELL 9: PREPARE FINAL DATASETS
# ==========================================

# Create final copies
employee_master = employee_df.copy()
employee_exits = exit_df.copy()
monthly_history = monthly_df.copy()

# Sort datasets
employee_master = employee_master.sort_values(
    "Employee_ID"
).reset_index(drop=True)

employee_exits = employee_exits.sort_values(
    "Exit_Date"
).reset_index(drop=True)

monthly_history = monthly_history.sort_values(
    ["Employee_ID", "Observation_Date"]
).reset_index(drop=True)

# Check duplicate employee-month records
monthly_duplicates = monthly_history.duplicated(
    subset=["Employee_ID", "Observation_Date"]
).sum()

print("EMPLOYEE MASTER:", employee_master.shape)
print("EMPLOYEE EXITS:", employee_exits.shape)
print("MONTHLY HISTORY:", monthly_history.shape)

print("\nUnique employees:",
      employee_master["Employee_ID"].nunique())

print("Duplicate employee-month records:",
      monthly_duplicates)

print("\nDatasets prepared successfully.")
```

```python
# Define chronological split date
split_date = pd.Timestamp("2025-01-01")

# Create training dataset: January to December 2024
train_data = modeling_history_eligible[
    modeling_history_eligible["Observation_Date"] < split_date
].copy()

# Create testing dataset: January to June 2025
test_data = modeling_history_eligible[
    modeling_history_eligible["Observation_Date"] >= split_date
].copy()

# Separate predictors and target
# Employee_ID and Observation_Date are excluded from model predictors

features = [
    "Months_At_Company",
    "Tenure_Years",
    "Job_Satisfaction",
    "Work_Life_Balance",
    "Engagement_Score",
    "Manager_Relationship",
    "Workload_Score",
    "Overtime_Hours"
]

target = "Attrition_Within_6M"

X_train = train_data[features]
X_test = test_data[features]

y_train = train_data[target]
y_test = test_data[target]

# Display split summary
print("CHRONOLOGICAL TRAIN-TEST SPLIT")
print("=" * 55)

print("\nTRAINING DATA")
print("Records:", len(X_train))
print("Start:", train_data["Observation_Date"].min())
print("End:", train_data["Observation_Date"].max())
print("Positive attrition labels:", y_train.sum())

print("\nTESTING DATA")
print("Records:", len(X_test))
print("Start:", test_data["Observation_Date"].min())
print("End:", test_data["Observation_Date"].max())
print("Positive attrition labels:", y_test.sum())

print("\nPREDICTOR VARIABLES:", len(features))
print("Target variable:", target)

print("\nChronological split completed.")
```

```python
# Define the date when the model is assumed to be trained
split_date = pd.Timestamp("2025-01-01")

# Keep training observations whose full six-month outcome
# window had ended by the training date
training_cutoff = split_date - pd.DateOffset(months=6)

train_data = modeling_history_eligible[
    modeling_history_eligible["Observation_Date"] <= training_cutoff
].copy()

# Keep the future testing period
test_data = modeling_history_eligible[
    (modeling_history_eligible["Observation_Date"] >= split_date) &
    (modeling_history_eligible["Observation_Date"] <= pd.Timestamp("2025-06-30"))
].copy()

# Define predictors and target
features = [
    "Months_At_Company",
    "Tenure_Years",
    "Job_Satisfaction",
    "Work_Life_Balance",
    "Engagement_Score",
    "Manager_Relationship",
    "Workload_Score",
    "Overtime_Hours"
]

target = "Attrition_Within_6M"

X_train = train_data[features]
X_test = test_data[features]

y_train = train_data[target]
y_test = test_data[target]

# Display results
print("PURGED CHRONOLOGICAL TRAIN-TEST SPLIT")
print("=" * 55)

print("\nTRAINING DATA")
print("Records:", len(X_train))
print("Start:", train_data["Observation_Date"].min())
print("End:", train_data["Observation_Date"].max())
print("Positive labels:", y_train.sum())

print("\nTESTING DATA")
print("Records:", len(X_test))
print("Start:", test_data["Observation_Date"].min())
print("End:", test_data["Observation_Date"].max())
print("Positive labels:", y_test.sum())

print("\nEMBARGO PERIOD")
print("July–December 2024")

print("\nTRAINING AND TESTING DATA PREPARED.")
```

```python
# Import machine learning libraries

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report
)

# Build the Logistic Regression pipeline

logistic_model = Pipeline([
    ("scaler", StandardScaler()),
    ("classifier", LogisticRegression(
        class_weight="balanced",
        max_iter=1000,
        random_state=42
    ))
])

# Train the model
logistic_model.fit(X_train, y_train)

# Generate predictions
y_pred = logistic_model.predict(X_test)

# Generate probability scores
y_probability = logistic_model.predict_proba(X_test)[:, 1]

# Evaluate model performance
print("BASELINE LOGISTIC REGRESSION RESULTS")
print("=" * 55)

print("Accuracy:", round(accuracy_score(y_test, y_pred), 4))
print("Precision:", round(precision_score(y_test, y_pred, zero_division=0), 4))
print("Recall:", round(recall_score(y_test, y_pred, zero_division=0), 4))
print("F1-Score:", round(f1_score(y_test, y_pred, zero_division=0), 4))
print("ROC-AUC:", round(roc_auc_score(y_test, y_probability), 4))
print("PR-AUC:", round(average_precision_score(y_test, y_probability), 4))

print("\nCLASSIFICATION REPORT")
print(classification_report(
    y_test,
    y_pred,
    target_names=["No Attrition", "Attrition"],
    zero_division=0
))

print("\nCONFUSION MATRIX")
print(confusion_matrix(y_test, y_pred))
```

```python
# Compare attrition and non-attrition observations
# using training data only

feature_comparison = (
    train_data.groupby("Attrition_Within_6M")[features]
    .mean()
    .T
    .reset_index()
)

feature_comparison.columns = [
    "Feature",
    "No_Attrition_Mean",
    "Attrition_Mean"
]

feature_comparison["Difference"] = (
    feature_comparison["Attrition_Mean"]
    - feature_comparison["No_Attrition_Mean"]
).round(3)

print("FEATURE COMPARISON BY ATTRITION OUTCOME")
display(feature_comparison.round(3))

# Compare against a simple majority-class baseline
majority_prediction = np.zeros(len(y_test), dtype=int)

baseline_accuracy = accuracy_score(y_test, majority_prediction)
baseline_recall = recall_score(
    y_test, majority_prediction, zero_division=0
)

print("\nMAJORITY-CLASS BASELINE")
print("Accuracy:", round(baseline_accuracy, 4))
print("Recall:", round(baseline_recall, 4))

# Check how many positive predictions the Logistic Regression made
print("\nLOGISTIC REGRESSION PREDICTION SUMMARY")
print("Predicted attrition:", int(y_pred.sum()))
print("Actual attrition:", int(y_test.sum()))
print(
    "Percentage predicted as attrition:",
    round(y_pred.mean() * 100, 2), "%"
)
```
